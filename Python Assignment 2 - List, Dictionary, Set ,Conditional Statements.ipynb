# List (Creation, Modification and Access):

## 1. List Creation:

```python
age_list=[24,25,26,27,28]
print(age_list)
```

```python
name_list=["Anu","Arathi","Athira","Hanan","Manu"]
print(name_list)
```

## 2. List Operations / Modifications:

```python
name_list.append('Yazhini')
print(name_list)
```

```python
age_list.insert(2,30)
print(age_list)
```

```python
name_list.remove('Yazhini')
print(name_list)
```

```python
age_list.pop()
```

```python
print(age_list)
```

```python
age_list.extend([29,30,26])
print(age_list)
```

```python
age_list.sort(reverse=True)
print(age_list)
```

```python
max(age_list)
min(age_list)
sum(age_list)
print("max age:",max(age_list))
print("min age:",min(age_list))
print("sum age:",sum(age_list))
```

## 3. Accessing List Elements:

```python
print(name_list[0])
print(name_list[-1])
print(name_list[2:5])
print(name_list[::-1])
```

# Dictionary (Creation, Modification and Access):

```python
student_marks={"Anu":85,"Arathi":98,"Athira":62,"Hanan":75,"Manu":88}
print(student_marks)
```

```python
print(student_marks["Arathi"])
```

```python
student_marks["Janani"]=80
print(student_marks)
```

```python
student_marks["Arathi"]=82
print(student_marks)
```

```python
print("All keys:",student_marks.keys())
print("All values:",student_marks.values())
print("All key-value pairs:",student_marks.items())
```

# Sets (Operations):

```python
my_set={'a','e','i','o','u','a','a','i'}
print(my_set)
#The set is automatically removed theduplicates a (which appears 3 times)& i (which appears twice).
#The order of elemets also changed since set is unordered
```

```python
my_set[4]='s'
# since a set is unordered it doesnot support indexing or item assignment.
```

```python
set1 = {1, 3, 5, 7, 9}
print("set1;",set1)
set2 = {2, 3, 5, 8, 10}
print("set2:",set2)
print("Union:",set1.union(set2))
print("Intersection:",set1.intersection(set2))
```

## Operators & Conditional Statements :

```python
while True:
    score = int(input("Enter your score (0 to 10): "))

    if score < 0 or score > 10:
        print("Invalid score! Please enter a score between 0 and 10.")

    elif score > 7:
        print("Above Average: Excellent performance! Keep it up.")

    elif score >= 4:
        print("Average: Good effort! Keep practicing.")

    else:
        print("Below Average: Need to improve your performance.")

    choice = input("Do you want to enter another score? (yes/no): ")

    if choice.lower() != "yes":
        print("Program ended.")
        break
```
